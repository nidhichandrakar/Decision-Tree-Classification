```python
df.columns = ['Study', 'Attendance', 'Assignments', 'Revision', 'Pass']
```

```python
for i in df.columns:
  df[i] = np.where(df[i] == "Yes",True,False)
```

```python
df.head()
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion = 'entropy', random_state = 0)
classifier.fit(X, y)
```

```python
from sklearn import tree

plt.figure(figsize=(20,10))
tree.plot_tree(classifier, feature_names=['Study', 'Attendance', 'Assignments', 'Revision'], class_names=['Y', 'N'], filled=True)
plt.title('Decision Tree Visualization')
plt.show()
```
